```python
import statistics
from pathlib import Path

def calculate_stats_from_files(directory: str):
    """
    Finds all .txt files in a directory, reads the first line as an integer,
    and calculates the mean and standard deviation of these numbers.

    Args:
        directory (str): The path to the directory containing .txt files.

    Returns:
        A tuple of (mean, std_dev) or (None, None) if no valid data is found.
    """
    numbers = []
    # Iterate through all files ending with .txt in the given directory
    for file_path in Path(directory).glob('*.txt'):
        try:
            with open(file_path, 'r') as f:
                first_line = f.readline()
                if first_line:
                    numbers.append(float(first_line.strip()))
        except (ValueError, IOError) as e:
            # Skip files that are empty, have non-integer first lines, or can't be read
            print(f"Warning: Skipping {file_path.name}: {e}")

    # Calculate statistics if we have enough data
    if len(numbers) >= 2:
        mean = statistics.mean(numbers)
        std_dev = statistics.stdev(numbers)
        return (mean, std_dev)
    elif len(numbers) == 1:
        return (numbers[0], 0) # Mean is the number itself, std dev is 0
    else:
        return (None, None)

calculate_stats_from_files("/home/mg873uh/Projects_kb/data/finetuning_preproc/Task003_FOMO3")
```

```python
import numpy as np
from pathlib import Path

def calculate_log_transformed_stats_from_files(directory: str):
    """
    Finds all .txt files, reads the first line as a number, applies a log1p
    transform, and then calculates the mean and standard deviation.

    Args:
        directory (str): The path to the directory containing .txt files.

    Returns:
        A tuple of (mean, std_dev) of the log-transformed data or (None, None).
    """
    numbers = []
    for file_path in Path(directory).glob('*.txt'):
        try:
            with open(file_path, 'r') as f:
                first_line = f.readline()
                if first_line:
                    numbers.append(float(first_line.strip()))
        except (ValueError, IOError) as e:
            print(f"Warning: Skipping {file_path.name}: {e}")

    if len(numbers) < 1:
        return (None, None)

    # 1. Transform the data using log1p
    log_transformed_numbers = np.log1p(numbers)
    
    # 2. Calculate stats on the transformed data
    mean = np.mean(log_transformed_numbers)
    # Use ddof=1 to calculate the sample standard deviation, same as statistics.stdev
    std_dev = np.std(log_transformed_numbers, ddof=1) if len(log_transformed_numbers) > 1 else 0.0

    return (mean, std_dev)

# Example usage:
log_mean, log_std = calculate_log_transformed_stats_from_files(
    "/home/mg873uh/Projects_kb/data/finetuning_preproc/Task003_FOMO3"
)

if log_mean is not None:
    print(f"Mean of log-transformed ages: {log_mean}")
    print(f"Std Dev of log-transformed ages: {log_std}")
```
